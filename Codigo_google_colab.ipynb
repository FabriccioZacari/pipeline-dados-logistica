```python
import pandas as pd
import numpy as np


np.random.seed(42)
n = 300  # 300 pedidos simulados

dados_logistica = {
    'id_pedido': [f'PED-{1000+i}' for i in range(n)],
    'transportadora': np.random.choice(['LogFast Brasil', 'TransRápido', 'ExpressCargo', 'RotaSegura'], n),
    'regiao': np.random.choice(['Sudeste', 'Sul', 'Nordeste', 'Centro-Oeste', 'Norte'], n),
    'peso_kg': np.round(np.random.uniform(1.0, 35.0, n), 2),
    'custo_frete': np.round(np.random.uniform(25.0, 300.0, n), 2),
    'dias_previstos': np.random.randint(3, 10, n),
    'dias_reais': np.random.randint(2, 14, n),
    'status_entrega': np.random.choice(['Entregue', 'Atrasado', 'Extraviado'], n, p=[0.75, 0.20, 0.05])
}


df_bruto = pd.DataFrame(dados_logistica)


print("--- DADOS BRUTOS DE LOGÍSTICA ---")
display(df_bruto.head(5))
```

```python

nome_arquivo = 'dados_logistica_tratados.csv'


df_bruto.to_csv(nome_arquivo, index=False, encoding='utf-8-sig')


from google.colab import files
files.download(nome_arquivo)
```
